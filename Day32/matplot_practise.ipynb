```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("resources/customer_data.csv")
df.columns
```

```python
df['gender'].value_counts().plot(kind='bar', color=['pink', 'blue'])
plt.title("Number of Customers by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
plt.hist(df['age'], bins=10, color='skyblue', edgecolor='black')
plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
df['payment_method'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Payment Method Distribution")
plt.ylabel("")  # remove extra label
plt.show()
```

```python
df.boxplot(column='age', by='gender')
plt.title("Age Distribution by Gender")
plt.suptitle("")  # removes the automatic "Boxplot grouped by gender"
plt.xlabel("Gender")
plt.ylabel("Age")
plt.show()
```

###  BroadCasting
- Broadcasting allows numpy to perform Operations on arrays with different shapes by virtually expanding dimensions so they match the larger array's shape
> Rules
- The dimensions have the same size(from right to left)
- OR, one of the dimensions have the size of 1

```python
arr1 = np.array([[1,2,3,4,5]])
arr2 = np.array([[1],[2],[3],[4],[5]])
arr1.shape
arr2.shape
arr1 * arr2
```

```python
# Example 1: (3,2) + (2,)
a = np.ones((3,2))
b = np.array([1,2])
print((a+b).shape)  # (3,2) ✅

# Example 2: (4,3) + (3,)
a = np.ones((4,3))
b = np.array([1,2,3])
print((a+b).shape)  # (4,3) ✅

# Example 3: (2,3) + (2,1)
a = np.ones((2,3))
b = np.array([[1],[2]])
print((a+b).shape)  # (2,3) ✅

# Example 4: (3,2) + (3,) 
a = np.ones((3,2))
b = np.array([1,2,3])
print(a+b)  # ERROR ❌
```
